```python
#| label: fetch-data
#| include: false

# ── DATA DOWNLOAD ────────────────────────────────────────────────────────────
# Pre-step: fetches the data sources so nothing has to be stored in GitHub.
# Files that are already present are reused, so re-running is cheap.
from pathlib import Path
from urllib.request import urlretrieve
from zipfile import ZipFile, is_zipfile

TRIP_DATA_URL = "https://s3.amazonaws.com/tripdata/202601-citibike-tripdata.zip"
SEPTEMBER_TRIP_DATA_URL = "https://s3.amazonaws.com/tripdata/202609-citibike-tripdata.zip"
STATION_STATUS_URL = "https://gbfs.lyft.com/gbfs/1.1/bkn/en/station_status.json"
STATION_INFO_URL = "https://gbfs.lyft.com/gbfs/1.1/bkn/en/station_information.json"

DATA_DIR = Path("datasource")
DATA_DIR.mkdir(parents=True, exist_ok=True)
TRIP_ARCHIVE = DATA_DIR / "202601-citibike-tripdata.zip"
TRIP_DIR = DATA_DIR / "202601-citibike-tripdata"
SEPTEMBER_TRIP_ARCHIVE = DATA_DIR / "202609-citibike-tripdata.zip"
SEPTEMBER_TRIP_DIR = DATA_DIR / "202609-citibike-tripdata"
STATION_STATUS_JSON = DATA_DIR / "station_status.json"
STATION_INFO_JSON = DATA_DIR / "station_information.json"

trip_sources = [
    (TRIP_DATA_URL, TRIP_ARCHIVE, TRIP_DIR),
    (SEPTEMBER_TRIP_DATA_URL, SEPTEMBER_TRIP_ARCHIVE, SEPTEMBER_TRIP_DIR),
]
for url, trip_archive, trip_dir in trip_sources:
    if not trip_archive.exists() or not is_zipfile(trip_archive):
        print("Downloading", trip_archive.name, flush=True)
        partial_archive = trip_archive.with_suffix(".zip.part")
        urlretrieve(url, partial_archive)
        if not is_zipfile(partial_archive):
            raise ValueError(f"Downloaded file is not a valid ZIP: {partial_archive}")
        partial_archive.replace(trip_archive)
    print("Trip archive:", trip_archive, flush=True)

    # Reuse extracted files so rerunning does not unzip the same data again.
    with ZipFile(trip_archive) as archive:
        for member in archive.infolist():
            extracted = trip_dir / member.filename
            if member.is_dir():
                continue
            if not extracted.exists() or extracted.stat().st_size != member.file_size:
                archive.extract(member, trip_dir)
    print("Trip CSVs:", ", ".join(sorted(p.name for p in trip_dir.rglob("*.csv"))), flush=True)

# Reuse saved GBFS snapshots; delete a JSON file to fetch a fresh snapshot.
for url, target in [(STATION_STATUS_URL, STATION_STATUS_JSON),
                    (STATION_INFO_URL, STATION_INFO_JSON)]:
    if target.exists():
        print("Using existing", target)
    else:
        urlretrieve(url, target)
        print("Saved", target)
```

## Problem and Motivation

When the subway stops, riders find another way. In Greenpoint, that way is often Citi Bike.

The G train is the only subway line in Greenpoint. In summer 2024 the MTA suspended it in three phases, from June 28 to September 3, to replace a signal system more than 90 years old [1]. Citi Bike usage at 21 nearby stations rose by nearly 14 percent [2]. The work is not over. By August 2026 the MTA had suspended parts of the G on 16 weekends that year [2]. Riders reported up to an hour looking for an open dock, and Lyft raised its rebalancing capacity by 20 percent [3]. More closures are set for December 2026 [4, 5].

@fig-full-docks shows the pattern in our own data. Between noon and 5 pm, Greenpoint stations had no open dock about three times as often on weekends as on weekdays: 12.6 percent against 4.3 percent. The largest swings, though, are the ordinary daily rhythm, so a shutdown effect has to be measured against it. The figure is descriptive and does not separate shutdown weekends from normal ones.

The response so far is reactive. Citi Bike has no estimate of how many extra trips a shutdown creates, at which stations, or for how long. Studies of rail closures in Washington, D.C. and London show the effect is measurable [6, 7]. We propose to build that estimate, starting with the G train, and to forecast the surge when the next shutdown is announced.

## 1. Machine Learning Problem and Use of Outputs

**Problem type.** Supervised regression on time series data, used for causal inference. The target is the hourly count of departures and arrivals at each Citi Bike station.

**How it works.** The model learns normal demand from the hour, day of week, weather, and each station's own pattern, using only periods with no nearby shutdown. During a shutdown it predicts the counterfactual: the demand we would have seen if the trains had run. The causal effect is observed minus predicted demand.

**Outputs.** A causal estimate for each past shutdown, in extra trips per station and hour. A surge forecast for each announced shutdown: which stations will fill or empty, and by how much.

**Who uses them.** Citi Bike can move bikes and free docks before a shutdown starts. DOT can place new docks where shutdowns hit hardest, and the MTA can plan shuttle service with bike demand in mind. Riders can see ahead of time which stations will have a bike or an open dock.

## 2. Data Sources

We use four sources the class provides and add two from the MTA, shown in bold.

| Source | Origin | Grain | Role |
|:---|:---|:---|:---|
| Station status | Class workspace, GBFS | Station, every 5 minutes | Full docks, live demand |
| Station information | Class workspace, GBFS | Station, daily | Location, capacity, ids |
| Trip history | BigQuery or Citi Bike [8] | One row per trip | Hourly demand |
| Weather | BigQuery | By hour or day | Rain and temperature |
| **MTA service alerts** | NY Open Data [9] | One row per alert update | Shutdown events |
| **MTA subway GTFS** | MTA [10][12] | One row per stop | Stop coordinates |

**New data.** The MTA publishes its service alerts since April 2020 as open data: about 525,000 alert updates, with the affected lines and the alert text [9]. The GTFS feed gives the coordinates of every subway stop [10][12]. Together they tell us which stops lost service and when. The open dataset lags: on October 4, 2026 it ended on August 12. So it serves as history. For current closures we read the live MTA alerts feed [10] and archive it daily, because the feed keeps no record of past alerts.

**Shared catalog table.** We will publish `mta_subway_shutdowns`: one row per shutdown event and affected stop, with the line, stop, coordinates, start and end time, and the Citi Bike stations within a quarter mile. Any team that forecasts demand can join it as a feature.

**Known challenge.** The alert date is when the notice was published, not when service stopped. The real window sits in the free text and has to be parsed. A hand built calendar of G closures checks the result.

## 3. Approach and Models

**Step 1: Baseline.** For each station and hour, the average demand at the same hour and day of the week over recent normal weeks. Every later model has to beat it.

**Step 2: Counterfactual model.** Because the target is a count, we start with **Poisson regression** [11] and move to **gradient boosted trees** with a Poisson objective. Predictors are hour, day of week, holidays, weather, station capacity, and each station's normal demand profile. The model trains only on station hours with no shutdown nearby and with the station in service. It gets no recent demand lags: yesterday's demand during a shutdown already contains the shutdown, so a lag would absorb the effect we want to measure.

**Step 3: Effect and surge.** A station counts as affected if it is within a quarter mile of a subway stop that lost service [6]. The shutdown effect is observed demand minus counterfactual demand, with an interval taken from the model's errors on normal days. Effects summarized by distance to the closed stop, hour, and day type become the surge forecast for the next announced shutdown.

## 4. Evaluation

A causal effect has no ground truth: we never observe the weekend where the train ran.

**Forecast accuracy.** On normal periods the truth exists. We report mean absolute error in trips per station hour and Poisson deviance, against the baseline. Splits are by time, never random.

**Causal validity.** Stations far from the closed line, and fake shutdowns placed on normal weekends, should both show no effect. The 2024 G shutdown should show a clear increase near the closed stops, in line with the reported 14 percent [2] and with a difference in differences estimate on the same event. Displaced riders may spill into nearby stations, so we test several distance bands.

**Surge forecast.** For each shutdown we compare the forecast made before it with the effect measured after it. We also check that the stations we flag are the ones that fill up, using minutes at zero open docks.

**In production.** From November 16 to December 14 a daily job logs forecast error on normal station hours. The G closures scheduled for December weekends [4, 5] give a live test: forecast before, measure after.

## 5. Training: Batch

Training will be batch, on a schedule. The causal design requires it: the model must never learn from shutdown hours, and an online learner would absorb each surge as it happens.

Four jobs run in the class workspace with no manual steps: alert ingestion (hourly), scoring of the previous day (daily), a surge forecast for each new planned shutdown, and retraining (weekly). A retrained model replaces the current one only if it does better on the most recent normal week.

Trip files arrive after the month ends, so daily scoring uses demand derived from station status: drops in bikes available count as departures, rises as arrivals. Trip history trains the model and checks that proxy.

## 6. Computational Needs

No large needs. Everything runs on the class workspace. Station status is about 725,000 rows a day, or 22 million rows a year once aggregated by station and hour. Poisson regression and gradient boosted trees train on that in minutes on CPUs. No GPUs or large neural networks are planned.

**One request.** MTA alerts are free text. If the workspace has a hosted LLM, we would use it to turn each alert into a structured row: line, stops, start, and end. The volume is small. If not, we parse with rules.

## 7. Minimum Viable Product

The MVP is the whole pipeline, end to end, for one line: the G train. Every piece is simple.

1. **Shutdown table.** G train closures with affected stops, times, and nearby Citi Bike stations.
2. **Demand table.** Hourly departures and arrivals per station, G corridor plus control stations.
3. **Model.** The baseline and a Poisson regression, trained on normal hours only.
4. **Effect estimate.** Observed minus counterfactual for each closure in our data, with an interval.
5. **Daily job.** Runs in the class workspace, scores the previous day, and logs the error.

**The demo** shows one closure weekend in Greenpoint: observed against counterfactual demand by hour, and the stations that surged most. Boosted trees, other lines, and the surge forecast come later.

**Main risks.** Parsing alert text. Separating real trips from rebalancing. Few normal weekends to learn from. A gap in closure dates, since the alert history ends before our status data begins.
